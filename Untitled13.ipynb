```python
import torch
torch.set_default_dtype(torch.float64)
# n is the timesteps
# p is how many t-p tokens we are seeing (recent window)
# g gist, a anchor, f frame tokens per step, c action

#each token can be identified by its block and kind

FRAME, GIST, ACT = 0, 1, 2
N, P, G, A, F, C = 12, 4, 2, 2, 3, 2
B = F + G + C          # tokens per step = 7
L = N * B              # total tokens = 84

kind_one_step = torch.tensor([FRAME]*F + [GIST]*G + [ACT]*C)
# repeating and adding list [0 0 0 1 1 2 2]
kind = kind_one_step.repeat(N)                  # (L,)
blk  = torch.arange(N).repeat_interleave(B)     # (L,)

print("blk")
print(blk[:14]) #lets see till 14 (two time steps)
print("kind")
print(kind[:14])
```

```python
qb = blk[:, None]    # shape (L, 1): query's step, as a column
kb = blk[None, :]    # shape (1, L): key's step, as a row
qk = kind[:, None]   # query's kind
kk = kind[None, :]   # key's kind

same = (qb == kb)    # query and key in the same step
past = (kb < qb)     # key is in an earlier step


```

```python
FRAME, GIST, ACT = 0, 1, 2

def check_mask(M, blk, kind, A, P):
    # blk[i] = block of token i, kind[i] in {FRAME, GIST, ACT}
    qb, kb = blk[:, None], blk[None, :]
    kk = kind[None, :]
    past = kb < qb

    assert M.any(dim=1).all(), "some row has no True (softmax -> NaN)"
    assert not (M & (kb > qb)).any(), "future token visible"
    assert not (M & past & (kk == ACT)).any(), "past action visible"
    assert (M | ~(past & (kk == GIST))).all(), "past gist hidden"

    keep_frame = past & (kk == FRAME) & ((kb < A) | (qb - kb < P))
    drop_frame = past & (kk == FRAME) & ~keep_frame
    assert (M | ~keep_frame).all(), "anchor/recent frame hidden"
    assert not (M & drop_frame).any(), "evicted frame visible"
    print("mask checks passed")
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import FancyBboxPatch

def plot_mask_paper(M, blk, kind, fname="memorywam_mask.png",
                    gap=0.14, fontsize=12):
    m = M.cpu().numpy()
    k = kind.cpu().numpy()
    b = blk.cpu().numpy()
    L = m.shape[0]

    on = {0: "#f5c6aa", 1: "#b5e3a0", 2: "#97d9f5"}   # f, g, a rows
    off = "#efefef"

    size = min(max(4, L * 0.45), 14)
    fig, ax = plt.subplots(figsize=(size, size))

    for i in range(L):
        for j in range(L):
            color = on[int(k[i])] if m[i, j] else off
            ax.add_patch(FancyBboxPatch(
                (j + gap / 2, i + gap / 2), 1 - gap, 1 - gap,
                boxstyle="round,pad=0,rounding_size=0.15",
                fc=color, ec="none"))

    ax.set_xlim(0, L)
    ax.set_ylim(L, 0)
    ax.set_aspect("equal")

    # one label per run of tokens with the same (step, kind)
    names = {0: "f", 1: "g", 2: "a"}
    centers, labels = [], []
    start = 0
    for i in range(1, L + 1):
        if i == L or (b[i], k[i]) != (b[start], k[start]):
            centers.append((start + i) / 2)
            labels.append(f"${names[int(k[start])]}_{{{int(b[start])}}}$")
            start = i

    ax.set_xticks(centers)
    ax.set_xticklabels(labels, fontsize=fontsize)
    ax.set_yticks(centers)
    ax.set_yticklabels(labels, fontsize=fontsize)
    ax.xaxis.tick_top()
    ax.tick_params(length=0)
    for sp in ax.spines.values():
        sp.set_visible(False)
    ax.set_xlabel("K", fontsize=fontsize)
    ax.xaxis.set_label_position("top")
    ax.set_ylabel("Q", fontsize=fontsize)

    plt.tight_layout()
    plt.savefig(fname, dpi=200, bbox_inches="tight")
    plt.show()
```

```python
# 1: own step.
# key is f or g -> every query in the step sees it
# key is a      -> only a queries see it
within     = same & ((kk != ACT) | (qk == ACT))

# 2: past gists are always visible
past_gist = past & (kk==FRAME | GIST )

# 3: past frames are visible if anchor OR recent
past_frame = past & (kk == FRAME) & ((kb < A) | (qb - kb < P))
#kb<2 and qb - kb <4
# Past actions: no rule needed
M = within | past_gist | past_frame
print(M[0].sum(), M[77].sum(), M[-1].sum())
check_mask(M, blk, kind, A, P)
plot_mask_paper(M, blk, kind, fname="mask_toy.png", fontsize=7)
```

```python
d = 16
torch.manual_seed(0)
x = torch.randn(L, d)

class Layer(torch.nn.Module):
    def __init__(self, d):
        super().__init__()
        self.q = torch.nn.Linear(d, d)
        self.k = torch.nn.Linear(d, d)
        self.v = torch.nn.Linear(d, d)

    def attend(self, q, k, v, mask):
        scores = q @ k.T / d**0.5
        scores = scores.masked_fill(~mask, float('-inf'))
        return torch.softmax(scores, dim=-1) @ v

#layers = [Layer(d)]   # 1 layer for now to test it
layers = [Layer(d) for _ in range(2)] # two layers

@torch.no_grad()
def full_pass(x):
    h = x
    for lyr in layers:
        q, k, v = lyr.q(h), lyr.k(h), lyr.v(h)
        h = h + lyr.attend(q, k, v, M)
    return h
```

```python
d = 16
caches = [dict(k=torch.empty(0, d),
               v=torch.empty(0, d),
               idx=torch.empty(0, dtype=torch.long))
          for _ in layers] #Allocate a 0xd tensor in memory

out_cached = torch.zeros_like(x)

with torch.no_grad():
    for t in range(N): #12 steps
        idx = torch.arange(t*B, (t+1)*B)   # global token ids of step t
        h = x[idx]                         # (7, 16)
        M_own = M[idx][:, idx]             # (7, 7)
        #current step (the latest frame, its gists, and its actions) looking at itself.


        for lyr, cache in zip(layers, caches): #zip([L1, L2], [c1, c2])  →  (L1, c1), (L2, c2)
            q, k, v = lyr.q(h), lyr.k(h), lyr.v(h)
            #1: keys, values = cache + this step
            keys   = torch.cat([cache["k"], k], dim=0) #keys: (n_cached + 7, 16)
            values = torch.cat([cache["v"], v], dim=0) # (0, 16) and (7, 16) → (7, 16) and so on (like keys)
            #2: mask = [all True for cache | M_own]
            m = M[idx][:, cache["idx"]]
            assert m.all(), f"step {t}: cache holds a key this step can't see"
            mask = torch.cat([m, M_own], dim=1) #mask: (7, n_cached + 7)
            #3: h = h + attend(...)
            h = h + lyr.attend(q, keys, values, mask)  #q @ keys.T =>(7, 16) @ (16, n_cached + 7)  →  (7, n_cached + 7)
            # softmax step =>(7, n_cached + 7) @ (n_cached + 7, 16)  →  (7, 16)
            #4: append frame+gist rows of k, v, idx to cache
            keep = kind[idx] != ACT
            cache["k"]   = torch.cat([cache["k"],   k[keep]],   dim=0)
            cache["v"]   = torch.cat([cache["v"],   v[keep]],   dim=0)
            cache["idx"] = torch.cat([cache["idx"], idx[keep]], dim=0)

        # TODO 5: evict frames of step t-(P-1) if it's >= A
        s = t - (P - 1)

        for cache in caches:
            ids  = cache["idx"]
            drop = (blk[ids] <= s) & (blk[ids] >= A) & (kind[ids] == FRAME) #what all drop
            cache["k"]   = cache["k"][~drop]
            cache["v"]   = cache["v"][~drop]
            cache["idx"] = cache["idx"][~drop]
        out_cached[idx] = h

print(torch.allclose(full_pass(x), out_cached, atol=1e-10))
ref = full_pass(x)
print(torch.allclose(ref, out_cached, atol=1e-10))
print("max diff:", (ref - out_cached).abs().max().item())
print("cache size:", caches[0]["k"].shape[0])
```
